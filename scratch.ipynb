```python
from transformers import AutoModelForCausalLM
model_id = "/root/cohere_ckpt/c3-7b/c3-7b-hf/hugging_face/poseidon"
model = AutoModelForCausalLM.from_pretrained(model_id)
```

```python
model.model.layers[0]
```

```python
import json
import pathlib

target_dir = "/root/cohere_ckpt/c3-7b/c3-7b-hf/hugging_face/sparse"
def generate_sparse_config(model):
    layer_config = {}
    for idx, _ in enumerate(model.model.layers):
        layer_config[idx] = [
            "vertical_and_slash",
            480,
            480,
            "0.95@128k",
        ]
    return layer_config

layer_config = generate_sparse_config(model)

with open(pathlib.Path(target_dir) / "sparse_attention_config.json", "w") as handle:
    layer_config_json = json.dumps(layer_config)
    handle.write(layer_config_json)
```
