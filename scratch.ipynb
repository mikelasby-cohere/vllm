```python
from transformers import AutoModelForCausalLM
model_id = "/root/cohere_ckpt/c3-7b-hf/hugging_face/poseidon"
# model_id = "/root/cohere_ckpt/sparse-r-32b/"
model = AutoModelForCausalLM.from_pretrained(model_id)
```

```python
model.model.layers[0].self_attn.k_proj.out_features/8
```

```python
import json
with open(f"{model_id}/config.json", "r") as handle:
    config = json.load(handle)
num_heads = config["num_attention_heads"]
```

```python
import json
import pathlib

target_dir = "/root/cohere_ckpt/c3-7b-hf/hugging_face/sparse"
def generate_sparse_config(model, num_heads):
    sparse_config = []

    for _ in model.model.layers:
        layer_config = {}
        for idx in range(num_heads):
            layer_config[idx] = [
                "vertical_and_slash",
                2048,
                2048,
                "0.80@128k",
            ]
        sparse_config.append(layer_config)
    return sparse_config

layer_config = generate_sparse_config(model, num_heads)

with open(pathlib.Path(target_dir) / "sparse_attention_config.json", "w") as handle:
    layer_config_json = json.dumps(layer_config)
    handle.write(layer_config_json)
```
